```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load all sheets into a dictionary
file_path = "C:/Users/wellw/Downloads/DATASET_SOLTRACK/SOLTRACK_January_2025.xlsx"
sheets = pd.read_excel(file_path, sheet_name=None, engine='openpyxl')
# Check sheet names
print(sheets.keys())
```

```python
# Combine all sheets into one DataFrame
df = pd.concat(sheets.values(), ignore_index=True)
#The data is split across multiple days (Jan 1, Jan 2, Jan 3, Jan 4). Each sheet is one day's worth of data.
#It's easier to work with one DataFrame instead of switching between sheets every time you want to analyze or visualize something.
# Check shape
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
df.head()
```

```python
# Drop Irradiance column (column E - index 4 in 0-based counting)
# First check column names
print(df.columns.tolist())

# Drop Irradiance
df = df.drop(columns=['Irradiance (W/m²)'])

# Verify removal
df.columns.tolist()
```
